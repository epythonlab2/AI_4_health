##**Introduction to Data Manipulation with Pandas**

####**What is Pandas?**

**Pandas** is Python’s module for working with **tabular** data (data that has **rows and columns**).

Pandas gives you the functionality of programs like **SQL** or **Excel** along with all the power of Python.

Data Analysts use Python to ingest, organize, and analyze large data sets.

By the end of the section, you will know
- how to ingest or create tables of data,
- summarize data with aggregates, and
- combine data from multiple tables using merge.

If you find yourself interested in exploring additional functionalities of Pandas as you complete the lessons in this section, check out the official [Pandas Documentation](https://pandas.pydata.org/).
